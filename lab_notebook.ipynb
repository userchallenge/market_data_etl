```python
# Database Query Setup
import pandas as pd
from market_data_etl.database.manager import DatabaseManager

pd.set_option("display.max_rows", 200)
# Initialize database connection
db = DatabaseManager()


def query_to_df(sql_query):
    """Execute SQL query and return results as DataFrame"""
    with db.get_session() as session:
        return pd.read_sql(sql_query, session.connection())


print("✅ Database connection ready")
```

```python
# Load all database tables as DataFrames
# Get all table names from the database
table_names = query_to_df("SELECT name FROM sqlite_master WHERE type='table'")[
    "name"
].tolist()

print(f"Found {len(table_names)} tables in database:")
print("Loading DataFrames...")

# Create DataFrame for each table
for table_name in table_names:
    df_name = f"df_{table_name}"
    df = query_to_df(f"SELECT * FROM {table_name}")

    # Make the DataFrame available in the notebook environment
    globals()[df_name] = df

    print(f"  {df_name}: {len(df)} rows, {len(df.columns)} columns")

print("\n✅ All tables loaded! Access them as: df_instruments, df_prices, etc.")
```

```python
# Create Non-Financial & Investment Analysis DataFrame
import numpy as np

# Filter for Traditional and Investment Company categories only
non_financial = final_enhanced_overview[
    final_enhanced_overview["company_category"].isin(
        ["Traditional", "Investment Company"]
    )
].copy()

# Select core columns
analysis_df = non_financial[
    [
        "ticker_symbol",
        "instrument_name",
        "latest_revenue",
        "latest_ebitda",
        "latest_fcf",
        "latest_price",
        "revenue_2022",
        "ebitda_2022",
        "fcf_2022",
        "price_2022",
        "median_revenue_growth",
        "median_ebitda_growth",
        "median_fcf_growth",
        "median_price_growth",
    ]
].copy()

# Calculate growth ratios (latest vs 2022)
analysis_df["latest_revenue_growth_vs_2022"] = (
    analysis_df["latest_revenue"] / analysis_df["revenue_2022"]
)
analysis_df["latest_ebitda_growth_vs_2022"] = (
    analysis_df["latest_ebitda"] / analysis_df["ebitda_2022"]
)
analysis_df["latest_fcf_growth_vs_2022"] = (
    analysis_df["latest_fcf"] / analysis_df["fcf_2022"]
)
analysis_df["latest_price_growth_vs_2022"] = (
    analysis_df["latest_price"] / analysis_df["price_2022"]
)

# Display the analysis DataFrame
analysis_df
```
